```python

# ==============================================================================
# STEP 1: Install Required Libraries
# ==============================================================================
!pip install --quiet qiskit qiskit-aer qiskit-ibm-runtime yfinance matplotlib pandas numpy

import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt

# Qiskit Imports
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# ==============================================================================
# STEP 2: Fetch Gold Market Data & Calculate Classical Metrics
# ==============================================================================
print("Fetching Gold market data...")

# Try tickers in order: Futures -> ETF 1 -> ETF 2
tickers = ["GC=F", "GLD", "IAU"]
gold_data = pd.DataFrame()

for ticker in tickers:
    # Use "1mo" for 1 month of daily data
    df = yf.Ticker(ticker).history(period="1mo")
    if not df.empty and len(df) > 1:
        gold_data = df
        print(f"Successfully fetched data using ticker: {ticker}")
        break

# Safety Fallback if Yahoo API fails completely
if gold_data.empty:
    print("Warning: Yahoo Finance API unreachable. Using simulated fallback price.")
    current_price = 2400.00
    volatility = 0.012
else:
    current_price = float(gold_data['Close'].iloc[-1])
    daily_returns = gold_data['Close'].pct_change().dropna()
    volatility = float(daily_returns.std()) if len(daily_returns) > 0 else 0.01

print(f"Current Price: ${current_price:.2f}")
print(f"Daily Volatility: {volatility*100:.2f}%\n")

# ==============================================================================
# STEP 3: Build Quantum Circuit
# ==============================================================================
num_qubits = 3
qc = QuantumCircuit(num_qubits, num_qubits)

# Parameterize rotation angle theta based on historical volatility
theta = 2 * np.arcsin(np.clip(volatility * 10, 0.05, 0.95))

qc.h(0)                   # Superposition
qc.ry(theta, 1)          # Volatility-driven rotation
qc.cx(0, 1)              # Entanglement
qc.ry(theta / 2, 2)
qc.cx(1, 2)

qc.measure(range(num_qubits), range(num_qubits))

# ==============================================================================
# STEP 4: Run Simulation
# ==============================================================================
backend = AerSimulator()
shots = 4096
job = backend.run(qc, shots=shots)
result = job.result()
counts = result.get_counts()

# ==============================================================================
# STEP 5: Map Bitstrings to Percentage Support & Resistance Levels
# ==============================================================================
levels_config = {
    '000': ('S3 (Strong Support)',      -0.0382),
    '001': ('S2 (Moderate Support)',    -0.0236),
    '010': ('S1 (Weak Support)',        -0.0100),
    '011': ('Pivot Point',               0.0000),
    '100': ('R1 (Weak Resistance)',      0.0100),
    '101': ('R2 (Moderate Resistance)',  0.0236),
    '110': ('R3 (Strong Resistance)',    0.0382),
    '111': ('R4 (Extended Resistance)',  0.0618)
}

print("="*72)
print("             QUANTUM SIMULATION RESULTS FOR GOLD")
print("="*72)
print(f"{'State':<7} | {'Level Type':<25} | {'Shift %':<10} | {'Price Target':<12} | {'Prob %'}")
print("-" * 72)

plot_data = []

for bitstring, (label, pct_shift) in levels_config.items():
    prob = counts.get(bitstring, 0) / shots
    target_price = current_price * (1 + pct_shift)
    plot_data.append((label, target_price, pct_shift * 100))

    print(f"{bitstring:<7} | {label:<25} | {pct_shift*100:>+6.2f}%    | ${target_price:>9.2f}   | {prob*100:>5.1f}%")

# ==============================================================================
# STEP 6: Visualization
# ==============================================================================
plt.figure(figsize=(10, 6))
if not gold_data.empty:
    plt.plot(gold_data.index, gold_data['Close'], label='Price History', color='gold', linewidth=2.5)
else:
    plt.axhline(y=current_price, label='Estimated Spot Price', color='gold', linewidth=2.5)

colors = ['darkred', 'red', 'orange', 'gray', 'lightgreen', 'green', 'darkgreen', 'purple']
for idx, (label, target_price, pct) in enumerate(plot_data):
    plt.axhline(y=target_price, color=colors[idx], linestyle='--', alpha=0.7, label=f"{label} ({pct:+.2f}%)")

plt.title("Gold Price & Quantum Support/Resistance Levels")
plt.xlabel("Date")
plt.ylabel("Price (USD)")
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
